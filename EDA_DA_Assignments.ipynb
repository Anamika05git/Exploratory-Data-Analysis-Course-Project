# DA 1 — Dataframe Operations and Data Transformations

```python

import pandas as pd
import numpy as np

# ==================================================
# PART A: CREATE THREE DATAFRAMES
# ==================================================

# Dataframe 1: ProductID, NameofProd, Price
data1 = {
    "ProductID": list(range(1001, 1011)),
    "NameofProd": [
        "Laptop", "Mouse", "Keyboard", "Monitor", "Printer",
        "Scanner", "Tablet", "Webcam", "Speaker", "Router"
    ],
    "Price": [55000, 500, 1200, 12000, 8000,
              4500, 25000, 1800, 3000, 2200]
}

dataframe1 = pd.DataFrame(data1)

# Dataframe 2: ProductID, Company_Location
data2 = {
    "ProductID": list(range(1006, 1016)),
    "Company_Location": [
        "Chennai", "Mumbai", "Delhi", "Pune", "Kochi",
        "Bengaluru", "Hyderabad", "Jaipur", "Kolkata", "Coimbatore"
    ]
}

dataframe2 = pd.DataFrame(data2)

# Dataframe 3: ProductID, NameofProd, Price
data3 = {
    "ProductID": list(range(1011, 1021)),
    "NameofProd": [
        "Microphone", "Headphones", "Projector", "Hard Drive",
        "SSD", "USB Drive", "Power Bank", "Smartwatch",
        "Camera", "Gamepad"
    ],
    "Price": [2500, 1800, 30000, 5000, 6500,
              700, 1500, 4500, 35000, 1200]
}

dataframe3 = pd.DataFrame(data3)

print("DATAFRAME 1")
print(dataframe1)

print("\nDATAFRAME 2")
print(dataframe2)

print("\nDATAFRAME 3")
print(dataframe3)


# ==================================================
# PART B: APPEND AND CONCATENATE
# ==================================================

# Append records of dataframe1 and dataframe3
appended_df = pd.concat(
    [dataframe1, dataframe3], ignore_index=True
)

print("\nAPPENDED DATAFRAME")
print(appended_df)

# Concatenate the two dataframes row-wise
concatenated_df = pd.concat(
    [dataframe1, dataframe3], axis=0, ignore_index=True
)

print("\nCONCATENATED DATAFRAME")
print(concatenated_df)


# ==================================================
# PART C: MERGE USING ProductID
# ==================================================

merged_df = pd.merge(
    dataframe1, dataframe2, on="ProductID", how="inner"
)

print("\nMERGED DATAFRAME")
print(merged_df)


# ==================================================
# PART D: JOIN OPERATIONS
# ==================================================

# Inner join: only matching ProductIDs
inner_join = pd.merge(
    dataframe1, dataframe2, on="ProductID", how="inner"
)
print("\nINNER JOIN")
print(inner_join)

# Left join: all rows from dataframe1
left_join = pd.merge(
    dataframe1, dataframe2, on="ProductID", how="left"
)
print("\nLEFT JOIN")
print(left_join)

# Right join: all rows from dataframe2
right_join = pd.merge(
    dataframe1, dataframe2, on="ProductID", how="right"
)
print("\nRIGHT JOIN")
print(right_join)

# Full outer join: all rows from both dataframes
full_outer_join = pd.merge(
    dataframe1, dataframe2, on="ProductID", how="outer"
)
print("\nFULL OUTER JOIN")
print(full_outer_join)


# ==================================================
# PART E: CREATE A NUMERIC VECTOR
# ==================================================


anamika = np.array([
    275, 25, 40, 55, 60,
    75, 80, 90, 100, 120
], dtype=float)

print("\nORIGINAL NUMERIC VECTOR")
print(anamika)


# ==================================================
# PART F: NORMALIZATION TO DIFFERENT RANGES
# ==================================================

minimum = anamika.min()
maximum = anamika.max()

# Normalize to [-3, 3]
norm_3 = -3 + (
    (anamika - minimum) / (maximum - minimum)
) * 6

print("\nNORMALIZATION TO [-3, 3]")
print(norm_3)

# Normalize to [-1, 1]
norm_1 = -1 + (
    (anamika - minimum) / (maximum - minimum)
) * 2

print("\nNORMALIZATION TO [-1, 1]")
print(norm_1)

# Min-max normalization to [0, 1]
minmax_norm = (
    (anamika - minimum) / (maximum - minimum)
)

print("\nMIN-MAX NORMALIZATION TO [0, 1]")
print(minmax_norm)


# ==================================================
# PART G: Z-SCORE NORMALIZATION
# ==================================================

z_score = (
    (anamika - anamika.mean()) / anamika.std(ddof=0)
)

print("\nZ-SCORE NORMALIZATION")
print(z_score)


# ==================================================
# PART H: DECIMAL SCALING NORMALIZATION
# ==================================================

j = len(str(int(np.max(np.abs(anamika)))))

decimal_scaled = anamika / (10 ** j)

print("\nDECIMAL SCALING NORMALIZATION")
print(decimal_scaled)
```


# DA 2 — Data Import, Transformation and Basic Statistical Analysis


```python

import pandas as pd
import numpy as np

# Load the dataset shared by faculty
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/indo_rct.csv"

df_original = pd.read_csv(url)

# Create a working copy for DA 2
df = df_original.copy()

# 1. Display the first 10 records
print("FIRST 10 RECORDS")
display(df.head(10))

# 2. Display the last 10 records
print("\nLAST 10 RECORDS")
display(df.tail(10))

# 3. Find the number of rows and columns
print("\nDATASET DIMENSIONS")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 4. Display column names
print("\nCOLUMN NAMES")
print(df.columns.tolist())

# 5. Display data types
print("\nDATA TYPES")
print(df.dtypes)

# 6. Display summary statistics
print("\nSUMMARY STATISTICS")
display(df.describe(include="all").T)

# 7. Count missing values in each column
print("\nMISSING VALUES BY COLUMN")
print(df.isnull().sum())

# 8. Count missing values in the entire dataset
print("\nTOTAL MISSING VALUES")
print(df.isnull().sum().sum())
```

```python

# ==================================================
# DA 2: MISSING VALUES AND DATA CLEANING
# ==================================================

# 11. Identify rows containing at least one missing value
missing_rows = df[df.isnull().any(axis=1)]

print("ROWS CONTAINING MISSING VALUES:")
print("Number of such rows:", len(missing_rows))
display(missing_rows.head(10))


# 12. Remove rows containing missing values
df_dropna = df.dropna().copy()

print("\nDATASET AFTER REMOVING MISSING ROWS")
print("Original shape:", df.shape)
print("Cleaned shape:", df_dropna.shape)
display(df_dropna.head(10))


# 13. Replace missing values in a numerical column
# Use age for mean imputation
df_mean = df.copy()

mean_col = "age" if "age" in df.columns else df.select_dtypes(
    include="number"
).columns[0]

mean_value = df_mean[mean_col].mean()
missing_before = df_mean[mean_col].isnull().sum()

df_mean[mean_col] = df_mean[mean_col].fillna(mean_value)

print("\nMEAN IMPUTATION")
print("Column:", mean_col)
print("Mean:", mean_value)
print("Missing before:", missing_before)
print("Missing after:", df_mean[mean_col].isnull().sum())


# 14. Replace missing values in another numerical column
df_median = df.copy()

numeric_cols = df.select_dtypes(include="number").columns.tolist()

median_col = "risk" if "risk" in df.columns else next(
    (col for col in numeric_cols if col != mean_col),
    mean_col
)

median_value = df_median[median_col].median()
missing_before = df_median[median_col].isnull().sum()

df_median[median_col] = df_median[median_col].fillna(median_value)

print("\nMEDIAN IMPUTATION")
print("Column:", median_col)
print("Median:", median_value)
print("Missing before:", missing_before)
print("Missing after:", df_median[median_col].isnull().sum())


# 15. Replace categorical missing values using the mode
df_mode = df.copy()

categorical_cols = df_mode.select_dtypes(
    include=["object", "category"]
).columns

for col in categorical_cols:
    mode_value = df_mode[col].mode(dropna=True)

    if not mode_value.empty:
        df_mode[col] = df_mode[col].fillna(mode_value.iloc[0])

print("\nMODE IMPUTATION")
print("Remaining missing values in categorical columns:")
print(df_mode[categorical_cols].isnull().sum())


# 16. Forward fill: use the previous non-missing value
df_ffill = df.ffill()

print("\nFORWARD FILL")
print("Missing values remaining:", df_ffill.isnull().sum().sum())


# 17. Backward fill: use the next non-missing value
df_bfill = df.bfill()

print("\nBACKWARD FILL")
print("Missing values remaining:", df_bfill.isnull().sum().sum())


# 18. Replace remaining missing values with constants
df_constant = df.copy()

for col in df_constant.columns:
    if df_constant[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df_constant[col]):
            df_constant[col] = df_constant[col].fillna(0)
        else:
            df_constant[col] = df_constant[col].fillna("Unknown")

print("\nCONSTANT VALUE IMPUTATION")
print("Remaining missing values:", df_constant.isnull().sum().sum())
display(df_constant.head(10))
```

```python

# ==================================================
# DA 2: REMAINING DATA TRANSFORMATIONS & STATISTICS
# Questions 19-28
# ==================================================

# Create a separate working copy
df2 = df.copy()

# Remove the extra row-number column if present
# It is an imported row index, not a clinical variable.
df2 = df2.drop(columns=["rownames"], errors="ignore")


# 19. Remove duplicate records
print("QUESTION 19: DUPLICATE RECORDS")

print("Duplicate rows found:", df2.duplicated().sum())

df2 = df2.drop_duplicates().copy()

print("Shape after removing duplicates:", df2.shape)


# 20. Convert appropriate variables into categorical types
print("\nQUESTION 20: CONVERT VARIABLES TO CATEGORY")

categorical_cols = df2.select_dtypes(
    include=["object", "category"]
).columns.tolist()

for col in categorical_cols:
    df2[col] = df2[col].astype("category")

print("Categorical columns:", categorical_cols)
print("\nUpdated data types:")
print(df2.dtypes)


# 21. Rename one or more column names
print("\nQUESTION 21: RENAME COLUMNS")

if "site" in df2.columns:
    df2 = df2.rename(columns={"site": "study_site"})

print("Renamed column: site -> study_site")
print(df2.columns.tolist())


# 22. Create a derived column using existing variables
print("\nQUESTION 22: CREATE A DERIVED COLUMN")

df2["age_group"] = pd.cut(
    pd.to_numeric(df2["age"], errors="coerce"),
    bins=[0, 29, 44, 59, np.inf],
    labels=["Up to 29", "30-44", "45-59", "60+"],
    include_lowest=True
)

print("Created age_group from age.")
display(df2[["age", "age_group"]].head(10))


# 23. Display summary statistics
print("\nQUESTION 23: SUMMARY STATISTICS")

display(df2.describe(include="all").T)


# 24. Find the minimum and maximum values
print("\nQUESTION 24: MINIMUM AND MAXIMUM")

age_values = pd.to_numeric(
    df2["age"], errors="coerce"
).dropna()

print("Minimum age:", age_values.min())
print("Maximum age:", age_values.max())


# 25. Calculate the range of a numerical variable
print("\nQUESTION 25: RANGE")

age_range = age_values.max() - age_values.min()

print("Age range:", age_range)


# 26. Calculate the variance
print("\nQUESTION 26: VARIANCE")

print("Age variance:", age_values.var())


# 27. Calculate the standard deviation
print("\nQUESTION 27: STANDARD DEVIATION")

print("Age standard deviation:", age_values.std())


# 28. Calculate the interquartile range (IQR)
print("\nQUESTION 28: INTERQUARTILE RANGE")

q1 = age_values.quantile(0.25)
q3 = age_values.quantile(0.75)
iqr = q3 - q1

print("First quartile (Q1):", q1)
print("Third quartile (Q3):", q3)
print("IQR:", iqr)


# Final preview of the transformed dataset
print("\nFINAL TRANSFORMED DATASET")
print("Shape:", df2.shape)
display(df2.head(10))
```

```python

# ============================================================
# DA 3: DATA CLEANING AND OUTLIER DETECTION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Work on a fresh copy of the original dataset
df3 = df_original.copy()

# Remove the imported row-number column, if present
df3 = df3.drop(columns=["rownames"], errors="ignore")


# 1. Display the first and last 6 records
print("FIRST 6 RECORDS")
display(df3.head(6))

print("\nLAST 6 RECORDS")
display(df3.tail(6))


# 2. Identify missing values and affected rows
print("\nMISSING VALUES BY COLUMN")
print(df3.isnull().sum())

missing_rows = df3[df3.isnull().any(axis=1)]

print("\nROWS WITH AT LEAST ONE MISSING VALUE")
print("Number of rows:", len(missing_rows))
display(missing_rows.head(10))


# 3. Remove columns with more than 70% missing values
missing_percentage = df3.isnull().mean() * 100

print("\nMISSING VALUE PERCENTAGE")
print(missing_percentage.round(2))

columns_to_remove = missing_percentage[
    missing_percentage > 70
].index.tolist()

print("\nCOLUMNS REMOVED (>70% MISSING):")
print(columns_to_remove)

df3 = df3.drop(columns=columns_to_remove)


# 4. Impute numeric values using mean and median
print("\nNUMERIC IMPUTATION")

if "age" in df3.columns:
    age_mean = df3["age"].mean()
    df3["age"] = df3["age"].fillna(age_mean)
    print("Age missing values filled using mean:", age_mean)

if "risk" in df3.columns:
    risk_median = df3["risk"].median()
    df3["risk"] = df3["risk"].fillna(risk_median)
    print("Risk missing values filled using median:", risk_median)


# Fill any other numeric columns using their mean
for col in df3.select_dtypes(include="number").columns:
    if df3[col].isnull().any():
        df3[col] = df3[col].fillna(df3[col].mean())


# 5. Impute categorical values using the mode
print("\nCATEGORICAL IMPUTATION")

for col in df3.select_dtypes(include=["object", "category"]).columns:
    mode_values = df3[col].mode(dropna=True)

    if df3[col].isnull().any() and not mode_values.empty:
        df3[col] = df3[col].fillna(mode_values.iloc[0])

print("Remaining missing values:", df3.isnull().sum().sum())


# 6. Identify and remove duplicate rows
print("\nDUPLICATE RECORDS")

print("Duplicates found:", df3.duplicated().sum())

df3 = df3.drop_duplicates().copy()

print("Shape after duplicate removal:", df3.shape)


# 7. Convert text values into lowercase and uppercase
print("\nTEXT CASE CONVERSION")

if "site" in df3.columns:
    df3["site_lower"] = df3["site"].astype(str).str.lower()
    df3["site_upper"] = df3["site"].astype(str).str.upper()

    display(df3[["site", "site_lower", "site_upper"]].head(10))
else:
    print("Site column unavailable; inspect the available text columns.")


# 8. Convert text columns to categorical data types
print("\nDATA TYPE CONVERSION")

text_columns = df3.select_dtypes(
    include=["object", "category"]
).columns.tolist()

for col in text_columns:
    df3[col] = df3[col].astype("category")

print(df3.dtypes)


# 9. Detect outliers in age using a boxplot and IQR
print("\nOUTLIER DETECTION USING AGE")

age_values = pd.to_numeric(
    df3["age"], errors="coerce"
)

Q1 = age_values.quantile(0.25)
Q3 = age_values.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outlier_mask = (
    (age_values < lower_limit) |
    (age_values > upper_limit)
)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of age outliers:", outlier_mask.sum())

plt.figure(figsize=(8, 4))
sns.boxplot(x=age_values)
plt.title("Boxplot of Age for Outlier Detection")
plt.xlabel("Age")
plt.show()


# 10. Remove age outliers into a separate dataframe
# Preserve missing ages, if any, rather than dropping them here.
keep_rows = age_values.between(
    lower_limit, upper_limit
) | age_values.isna()

df3_no_outliers = df3.loc[keep_rows].copy()

print("\nDATASET AFTER AGE OUTLIER REMOVAL")
print("Before:", df3.shape)
print("After:", df3_no_outliers.shape)
```


# DA 4 — Bivariate Analysis


```python

# ============================================================
# DA 4: BIVARIATE ANALYSIS — CATEGORICAL AND NUMERIC VARIABLES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load a fresh copy of the original dataset
da4 = df_original.copy()

# Remove ID/index columns from analysis where appropriate
da4 = da4.drop(columns=["rownames"], errors="ignore")

# ------------------------------------------------------------
# 1. Contingency table for two categorical variables
# ------------------------------------------------------------

print("CONTINGENCY TABLE: GENDER vs OUTCOME")

cat_data = da4.dropna(subset=["gender", "outcome"])

contingency_table = pd.crosstab(
    cat_data["gender"],
    cat_data["outcome"]
)

display(contingency_table)


# ------------------------------------------------------------
# 2. Stacked bar plot using pandas/matplotlib
# ------------------------------------------------------------

contingency_table.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title("Stacked Bar Plot: Gender vs Outcome")
plt.xlabel("Gender")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.legend(title="Outcome")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 3. Stacked bar plot using seaborn-style plotting
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=cat_data,
    x="gender",
    hue="outcome",
    multiple="stack",
    shrink=0.8
)

plt.title("Stacked Count Plot: Gender and Outcome")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 4. Grouped / side-by-side bar plot
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=cat_data,
    x="gender",
    hue="outcome"
)

plt.title("Grouped Bar Plot: Gender vs Outcome")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.legend(title="Outcome")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. Scatter plot for two numerical variables
# ------------------------------------------------------------

numeric_pair = da4[["age", "risk"]].apply(
    pd.to_numeric, errors="coerce"
).dropna()

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=numeric_pair,
    x="age",
    y="risk"
)

plt.title("Scatter Plot: Age vs Risk")
plt.xlabel("Age")
plt.ylabel("Risk")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. Line plot showing the trend in average risk by age
# ------------------------------------------------------------

average_risk = (
    numeric_pair.groupby("age", as_index=False)["risk"]
    .mean()
    .sort_values("age")
)

plt.figure(figsize=(8, 5))

plt.plot(
    average_risk["age"],
    average_risk["risk"],
    marker="o"
)

plt.title("Average Risk by Age")
plt.xlabel("Age")
plt.ylabel("Average Risk")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. Heatmap for numerical variables
# ------------------------------------------------------------

numeric_cols = [
    col for col in da4.select_dtypes(include=np.number).columns
    if col not in ["rownames", "id"]
]

# Keep columns that have at least two valid distinct values
numeric_cols = [
    col for col in numeric_cols
    if da4[col].nunique(dropna=True) > 1
]

correlation_data = da4[numeric_cols].corr()

print("NUMERICAL CORRELATION MATRIX")
display(correlation_data.round(3))

if len(numeric_cols) >= 2:
    plt.figure(figsize=(9, 6))

    sns.heatmap(
        correlation_data,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        vmin=-1,
        vmax=1
    )

    plt.title("Correlation Heatmap of Numerical Variables")
    plt.tight_layout()
    plt.show()
else:
    print("Not enough numerical variables for a correlation heatmap.")
```


# DA 6 — Time Series Analysis (AirPassengers)


```python

# ============================================================
# DA 6: TIME SERIES ANALYSIS — DATA, PLOT AND DECOMPOSITION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# Load the classic AirPassengers dataset
url = (
    "https://raw.githubusercontent.com/"
    "jbrownlee/Datasets/master/airline-passengers.csv"
)

air = pd.read_csv(url, parse_dates=["Month"])

# Set Month as the time-series index
air = air.set_index("Month")
air.index = pd.date_range(
    start=air.index[0],
    periods=len(air),
    freq="MS"
)

air_ts = air["Passengers"].astype(float)
air_ts.name = "Passengers"

# 1. Display the first and last five observations
print("FIRST 5 OBSERVATIONS")
display(air.head())

print("\nLAST 5 OBSERVATIONS")
display(air.tail())

# 2. Identify the start and end periods
print("\nTIME SERIES PERIOD")
print("Start:", air_ts.index.min())
print("End:", air_ts.index.max())

# 3. Check the number of observations and frequency
print("\nTIME SERIES INFORMATION")
print("Number of observations:", len(air_ts))
print("Frequency: Monthly")
print("Seasonal period: 12 months")

# 4. Plot the complete monthly passenger series
plt.figure(figsize=(12, 5))
plt.plot(air_ts, linewidth=1.5)
plt.title("Monthly International Airline Passengers")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (thousands)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 5. Decompose the series into trend, seasonal and residual
decomposition = seasonal_decompose(
    air_ts,
    model="multiplicative",
    period=12
)

fig = decomposition.plot()
fig.set_size_inches(12, 9)
fig.suptitle("AirPassengers Time Series Decomposition", y=1.02)
plt.tight_layout()
plt.show()

# 6. Plot the autocorrelation function (ACF)
fig, ax = plt.subplots(figsize=(10, 4))
plot_acf(air_ts, lags=40, ax=ax)
ax.set_title("Autocorrelation Function (ACF)")
plt.tight_layout()
plt.show()

# 7. Plot the partial autocorrelation function (PACF)
fig, ax = plt.subplots(figsize=(10, 4))
plot_pacf(air_ts, lags=40, method="ywm", ax=ax)
ax.set_title("Partial Autocorrelation Function (PACF)")
plt.tight_layout()
plt.show()

print("\nDA 6 CELL 1 COMPLETED")
```

```python

# ============================================================
# DA 6: SEASONAL ARIMA MODEL AND 12-MONTH FORECAST
# ============================================================

from statsmodels.tsa.statespace.sarimax import SARIMAX

# 8. Fit a seasonal ARIMA model
# ARIMA orders: (p, d, q) = (2, 1, 2)
# Seasonal orders: (P, D, Q, s) = (1, 1, 1, 12)

model = SARIMAX(
    air_ts,
    order=(2, 1, 2),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

model_fit = model.fit(disp=False)

# 9. Display model parameters and summary
print("FITTED MODEL PARAMETERS")
print(model_fit.params)

print("\nMODEL SUMMARY")
print(model_fit.summary())

# 10. Forecast the next 12 months
forecast_result = model_fit.get_forecast(steps=12)

forecast_mean = forecast_result.predicted_mean
forecast_ci = forecast_result.conf_int()

# Create a table of forecasts
forecast_table = pd.DataFrame({
    "Forecast_Passengers": forecast_mean,
    "Lower_95_CI": forecast_ci.iloc[:, 0],
    "Upper_95_CI": forecast_ci.iloc[:, 1]
})

forecast_table.index.name = "Month"

print("\n12-MONTH PASSENGER FORECAST")
display(forecast_table.round(2))

# 11. Plot historical observations and forecasts
plt.figure(figsize=(12, 6))

plt.plot(
    air_ts.index,
    air_ts,
    label="Historical Passengers"
)

plt.plot(
    forecast_mean.index,
    forecast_mean,
    label="Forecast",
    linestyle="--",
    marker="o"
)

plt.fill_between(
    forecast_mean.index,
    forecast_ci.iloc[:, 0].to_numpy(),
    forecast_ci.iloc[:, 1].to_numpy(),
    alpha=0.2,
    label="95% Confidence Interval"
)

plt.title("AirPassengers: 12-Month Seasonal ARIMA Forecast")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (thousands)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nDA 6 COMPLETED")
```


# DA 8 — K-Means and Hierarchical Clustering


```python

# ============================================================
# DA 8: K-MEANS CLUSTERING AND ELBOW METHOD
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Start with the original dataset
da8 = df_original.copy()

# Remove row IDs; they should not influence clustering
da8 = da8.drop(
    columns=["rownames", "id"],
    errors="ignore"
)

# Display target/dependent variable distribution if available
if "outcome" in da8.columns:
    print("OUTCOME DISTRIBUTION (DESCRIPTIVE ONLY)")
    display(da8["outcome"].value_counts(dropna=False).to_frame("Count"))


# Convert yes/no clinical indicators to 0/1
binary_features = []

for col in da8.columns:
    if col in ["outcome", "gender", "site", "status", "type", "rx"]:
        continue

    values = set(
        da8[col].dropna().astype(str).str.strip().unique()
    )

    if values and values.issubset({"0_no", "1_yes"}):
        da8[col] = (
            da8[col].astype("string").str.strip()
            .map({"0_no": 0, "1_yes": 1})
        )

        if da8[col].nunique(dropna=True) > 1:
            binary_features.append(col)


# Select numeric features, excluding outcome and ID fields
exclude = {"outcome", "id", "rownames"}

numeric_features = [
    col for col in da8.select_dtypes(include=np.number).columns
    if col not in exclude and da8[col].nunique(dropna=True) > 1
]

print("\nFEATURES SELECTED FOR CLUSTERING")
print(numeric_features)

# Prepare numeric data
X_raw = da8[numeric_features].replace(
    [np.inf, -np.inf], np.nan
).copy()

# Remove columns containing no usable values
X_raw = X_raw.dropna(axis=1, how="all")

# Impute missing numeric values with the median
imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(X_raw)

feature_names = X_raw.columns.tolist()

X_prepared = pd.DataFrame(
    X_imputed,
    columns=feature_names,
    index=X_raw.index
)

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_prepared)

print("\nDATA PREPARATION")
print("Rows:", X_scaled.shape[0])
print("Features:", X_scaled.shape[1])
print("Missing values after imputation:", np.isnan(X_scaled).sum())


# ------------------------------------------------------------
# 1. K-Means clustering with k = 2
# ------------------------------------------------------------

kmeans2 = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

labels_k2 = kmeans2.fit_predict(X_scaled)

kmeans_results = da8.loc[X_prepared.index].copy()
kmeans_results["KMeans_Cluster_k2"] = labels_k2 + 1

print("\nK-MEANS CLUSTER ASSIGNMENTS (k=2)")
display(kmeans_results.head(10))

print("\nSAMPLES IN EACH K-MEANS CLUSTER")
display(
    kmeans_results["KMeans_Cluster_k2"]
    .value_counts()
    .sort_index()
    .rename("Number of Samples")
    .to_frame()
)

# Cluster centers in standardized space
centers_scaled = pd.DataFrame(
    kmeans2.cluster_centers_,
    columns=feature_names,
    index=["Cluster 1", "Cluster 2"]
)

print("\nK-MEANS CLUSTER CENTERS (STANDARDIZED FEATURES)")
display(centers_scaled.round(3))


# ------------------------------------------------------------
# 2. Calculate WCSS / inertia for k = 1 to 10
# ------------------------------------------------------------

max_k = min(10, len(X_scaled))
wcss = []

for k in range(1, max_k + 1):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)
    wcss.append(model.inertia_)

elbow_table = pd.DataFrame({
    "Number of Clusters (k)": range(1, max_k + 1),
    "WCSS / Inertia": wcss
})

print("\nELBOW METHOD RESULTS")
display(elbow_table.round(3))


# ------------------------------------------------------------
# 3. Plot the elbow curve
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, max_k + 1),
    wcss,
    marker="o"
)

plt.title("Elbow Method for Choosing K")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS / Inertia")
plt.xticks(range(1, max_k + 1))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nDA 8 CELL 1 COMPLETED")
```

```python

# ============================================================
# DA 8: FINAL K-MEANS AND HIERARCHICAL CLUSTERING
# ============================================================

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster


# ------------------------------------------------------------
# 1. Refit K-Means using the selected number of clusters
# ------------------------------------------------------------

chosen_k = min(3, len(X_scaled))  # Adjust after inspecting elbow plot

kmeans_final = KMeans(
    n_clusters=chosen_k,
    random_state=42,
    n_init=10
)

final_labels = kmeans_final.fit_predict(X_scaled)

kmeans_final_results = da8.loc[X_prepared.index].copy()
kmeans_final_results["KMeans_Cluster"] = final_labels + 1

print("SELECTED NUMBER OF CLUSTERS:", chosen_k)

print("\nFINAL K-MEANS CLUSTER COUNTS")
display(
    kmeans_final_results["KMeans_Cluster"]
    .value_counts()
    .sort_index()
    .rename("Number of Samples")
    .to_frame()
)

print("\nFINAL K-MEANS ASSIGNMENTS")
display(kmeans_final_results.head(10))

centers_original = pd.DataFrame(
    scaler.inverse_transform(kmeans_final.cluster_centers_),
    columns=feature_names,
    index=[
        f"Cluster {i + 1}" for i in range(chosen_k)
    ]
)

print("\nCLUSTER CENTERS IN ORIGINAL FEATURE UNITS")
display(centers_original.round(3))


# ------------------------------------------------------------
# 2. Prepare data for hierarchical clustering
# Use a sample if the dataset is very large
# ------------------------------------------------------------

max_hierarchical_rows = 2000

if len(X_scaled) > max_hierarchical_rows:
    rng = np.random.default_rng(42)

    sample_positions = np.sort(
        rng.choice(
            len(X_scaled),
            size=max_hierarchical_rows,
            replace=False
        )
    )

    X_hier_scaled = X_scaled[sample_positions]
    hier_indices = X_prepared.index[sample_positions]

    print(
        "\nHierarchical clustering uses a reproducible sample of",
        max_hierarchical_rows,
        "rows to limit computation."
    )
else:
    X_hier_scaled = X_scaled
    hier_indices = X_prepared.index

print("Rows used for hierarchical clustering:", len(X_hier_scaled))


# ------------------------------------------------------------
# 3. Calculate Euclidean distances
# Display a small distance matrix for readability
# ------------------------------------------------------------

small_sample = X_hier_scaled[:min(5, len(X_hier_scaled))]

distance_matrix = squareform(
    pdist(small_sample, metric="euclidean")
)

print("\nEUCLIDEAN DISTANCE MATRIX FOR FIRST 5 SAMPLES")
display(pd.DataFrame(distance_matrix).round(3))


# ------------------------------------------------------------
# 4. Perform Ward hierarchical clustering
# ------------------------------------------------------------

ward_linkage = linkage(
    X_hier_scaled,
    method="ward",
    metric="euclidean"
)

print("\nWARD LINKAGE MATRIX (FIRST 5 MERGES)")
display(pd.DataFrame(
    ward_linkage[:5],
    columns=[
        "Cluster 1", "Cluster 2",
        "Distance", "Number of Samples"
    ]
).round(3))


# ------------------------------------------------------------
# 5. Plot the dendrogram
# Truncate the display so the plot remains readable
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

dendrogram(
    ward_linkage,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=45,
    leaf_font_size=9
)

plt.title("Hierarchical Clustering Dendrogram (Ward's Method)")
plt.xlabel("Merged Groups")
plt.ylabel("Ward Linkage Distance")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. Cut hierarchy into 3 clusters
# ------------------------------------------------------------

hier_labels = fcluster(
    ward_linkage,
    t=3,
    criterion="maxclust"
)

hier_results = da8.loc[hier_indices].copy()
hier_results["Hierarchical_Cluster"] = hier_labels

print("\nHIERARCHICAL CLUSTER ASSIGNMENTS")
display(hier_results.head(10))

print("\nNUMBER OF SAMPLES IN EACH HIERARCHICAL CLUSTER")
display(
    hier_results["Hierarchical_Cluster"]
    .value_counts()
    .sort_index()
    .rename("Number of Samples")
    .to_frame()
)

print("\nDA 8 COMPLETED")
```


# DA 9 — Principal Component Analysis (PCA)


```python

# ============================================================
# DA 9: PCA — DATA PREPARATION, SUMMARY AND SCREE PLOT
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Load the original dataset
da9 = df_original.copy()

# Remove row identifiers; they are not useful PCA features
da9 = da9.drop(
    columns=["rownames", "id"],
    errors="ignore"
)

print("DATASET DIMENSIONS")
print("Rows:", da9.shape[0])
print("Columns:", da9.shape[1])


# 2. Convert yes/no clinical features to 0/1
# Exclude outcome and categorical descriptors from the PCA features.
excluded = {
    "outcome", "gender", "site", "status",
    "type", "rx", "id", "rownames"
}

binary_features = []

for col in da9.columns:
    if col in excluded:
        continue

    values = set(
        da9[col].dropna().astype(str).str.strip().unique()
    )

    if values == {"0_no", "1_yes"}:
        da9[col] = (
            da9[col].astype("string").str.strip()
            .map({"0_no": 0, "1_yes": 1})
        )
        binary_features.append(col)


# 3. Select numeric features
numeric_features = [
    col for col in da9.select_dtypes(include=np.number).columns
    if col not in excluded
]

# Explicitly convert age and risk if necessary
for col in ["age", "risk"]:
    if col in da9.columns and col not in numeric_features:
        da9[col] = pd.to_numeric(da9[col], errors="coerce")
        numeric_features.append(col)

# Remove duplicate names and columns that are constant
numeric_features = list(dict.fromkeys(numeric_features))
numeric_features = [
    col for col in numeric_features
    if da9[col].nunique(dropna=True) > 1
]

print("\nNUMERIC FEATURES SELECTED FOR PCA")
print(numeric_features)

if len(numeric_features) < 2:
    raise ValueError(
        "PCA requires at least two usable numeric features. "
        "Inspect the dataset's columns and data types."
    )


# 4. Prepare data and handle missing values
X_raw = da9[numeric_features].replace(
    [np.inf, -np.inf], np.nan
).dropna(axis=1, how="all")

numeric_features = X_raw.columns.tolist()

imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(X_raw)

# 5. Standardize all selected features
scaler = StandardScaler()
data_scaled = scaler.fit_transform(X_imputed)

print("\nSTANDARDIZED DATA")
print("Shape:", data_scaled.shape)
print("Missing values:", np.isnan(data_scaled).sum())


# 6. Perform PCA
n_components = min(data_scaled.shape)

pca = PCA(n_components=n_components)
pca_result = pca.fit_transform(data_scaled)

# 7. Print PCA summary / explained variance
explained_variance = pca.explained_variance_ratio_ * 100
cumulative_variance = np.cumsum(explained_variance)

pca_summary = pd.DataFrame({
    "Principal Component": [
        f"PC{i+1}" for i in range(n_components)
    ],
    "Eigenvalue": pca.explained_variance_,
    "Explained Variance (%)": explained_variance,
    "Cumulative Variance (%)": cumulative_variance
})

print("\nPCA SUMMARY")
display(pca_summary.round(3))


# 8. Scree plot — y-axis limited to 0 to 60
plt.figure(figsize=(10, 5))

components = np.arange(1, n_components + 1)

plt.bar(
    components,
    explained_variance,
    edgecolor="black"
)

plt.plot(
    components,
    explained_variance,
    marker="o",
    linewidth=1
)

plt.title("Scree Plot of Principal Components")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance (%)")
plt.xticks(components)
plt.ylim(0, 60)
plt.grid(axis="y", alpha=0.3)

# Label percentages, keeping annotations inside the plotting range
for x, value in zip(components, explained_variance):
    y_position = min(value + 1, 57)
    plt.text(
        x, y_position,
        f"{value:.1f}%",
        ha="center",
        fontsize=8
    )

plt.tight_layout()
plt.show()

print("\nDA 9 CELL 1 COMPLETED")
```

```python

# ============================================================
# DA 9: PCA — BIPLOT, SCORES AND PC1 VS PC2
# ============================================================

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 1. Create the PCA scores dataframe
pca_scores = pd.DataFrame(
    pca_result,
    columns=[
        f"PC{i+1}" for i in range(pca_result.shape[1])
    ],
    index=X_raw.index
)

print("FIRST 6 ROWS OF PCA SCORES")
display(pca_scores.head(6))


# 2. Create the PCA biplot
# Arrows show feature loadings; dots show observations.

# Import the label-placement helper; install it if needed
try:
    from adjustText import adjust_text
except ImportError:
    import subprocess
    import sys

    try:
        subprocess.check_call([
            sys.executable, "-m", "pip",
            "install", "-q", "adjustText"
        ])
        from adjustText import adjust_text
    except Exception:
        adjust_text = None

scores_2d = pca_scores[["PC1", "PC2"]].to_numpy()

# Feature loadings for PC1 and PC2
loadings = pca.components_[:2, :].T

# Select the strongest contributing features to keep labels readable
strength = np.sqrt((loadings ** 2).sum(axis=1))
top_n = min(12, len(numeric_features))

selected_indices = np.argsort(strength)[-top_n:]

score_extent = max(
    np.max(np.abs(scores_2d)),
    1e-8
)

loading_extent = max(
    np.max(np.abs(loadings[selected_indices])),
    1e-8
)

arrow_scale = 0.7 * score_extent / loading_extent

fig, ax = plt.subplots(figsize=(11, 8))

ax.scatter(
    scores_2d[:, 0],
    scores_2d[:, 1],
    alpha=0.35,
    s=18,
    color="steelblue",
    label="Observations"
)

texts = []

for i in selected_indices:
    x_loading = loadings[i, 0] * arrow_scale
    y_loading = loadings[i, 1] * arrow_scale

    ax.annotate(
        "",
        xy=(x_loading, y_loading),
        xytext=(0, 0),
        arrowprops=dict(
            arrowstyle="->",
            color="darkred",
            lw=1.2
        )
    )

    texts.append(
        ax.text(
            x_loading,
            y_loading,
            numeric_features[i],
            fontsize=9
        )
    )

if adjust_text is not None:
    adjust_text(
        texts,
        ax=ax,
        arrowprops=dict(
            arrowstyle="-",
            color="gray",
            lw=0.5
        )
    )
else:
    # Fallback if label-adjustment installation is unavailable
    for i, txt in enumerate(texts):
        txt.set_position((
            txt.get_position()[0],
            txt.get_position()[1] + (i % 3) * 0.025 * score_extent
        ))

ax.axhline(0, color="gray", linewidth=0.7)
ax.axvline(0, color="gray", linewidth=0.7)

ax.set_title("PCA Biplot — First Two Principal Components")
ax.set_xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% variance)"
)
ax.set_ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% variance)"
)

ax.legend()
plt.tight_layout()
plt.show()


# 3. PC1 versus PC2 using blue circular points
plt.figure(figsize=(9, 6))

plt.scatter(
    pca_scores["PC1"],
    pca_scores["PC2"],
    color="blue",
    marker="o",
    alpha=0.65
)

plt.title("PCA Scores Plot — Registration Number: 0275")
plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% variance)"
)
plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% variance)"
)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nDA 9 COMPLETED")
```
